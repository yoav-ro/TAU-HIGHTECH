Imports:


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Import csv file:

```python
full_file_path=(r"C:\Users\yoavr\Documents\Dev\TAU-HIGHTECH\Year 1\Data-Science\Class\project_data.csv")
data_file= pd.read_csv(full_file_path)
```

Show Histograms and Box Plots for age, ejection fraction and serum sodium based on a count of 299 patients: 

```python

# Descriptive statistics
descriptive_stats = data_file[['age', 'ejection_fraction', 'serum_sodium']].describe()

# Plot histograms
plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
plt.hist(data_file['age'], bins=20, edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.subplot(1, 3, 2)
plt.hist(data_file['ejection_fraction'], bins=20, edgecolor='black')
plt.title('Ejection Fraction Distribution')
plt.xlabel('Ejection Fraction (%)')
plt.ylabel('Frequency')

plt.subplot(1, 3, 3)
plt.hist(data_file['serum_sodium'], bins=20, edgecolor='black')
plt.title('Serum Sodium Distribution')
plt.xlabel('Serum Sodium (mEq/L)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

# Plot boxplots
plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
sns.boxplot(y=data_file['age'])
plt.title('Age Boxplot')

plt.subplot(1, 3, 2)
sns.boxplot(y=data_file['ejection_fraction'])
plt.title('Ejection Fraction Boxplot')

plt.subplot(1, 3, 3)
sns.boxplot(y=data_file['serum_sodium'])
plt.title('Serum Sodium Boxplot')

plt.tight_layout()
plt.show()
```

From the histograms, we can see that execpt several outliers, the histogram has a shape familiar to a normal distribution, a common behavior in nature and fitting the Central Limit Theorem.
The outliers on these graphs might be attributed to varius health condition statistics of the patriens.
