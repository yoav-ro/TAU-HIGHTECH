Calculate the correlation matrix C, Then calculate the determinant and the inverse of C <br><br>
Reminder : <br>

g) $det(C) = det(X^TX)$ <br>

The function should return a **dictionary** containing the values:<br>
* C = Correlation matrix
* C_det = Determinant of C

```python
def corr_mat_atts(X):
  corr=np.matmul(X.T,X)
  return {'C':corr,'C_det':np.linalg.det(corr),'C_inv':np.linalg.inv(corr),}
```

## 2. Given a scalar and an 1d array, find the closest value to this scalar
Complete the function `farthest_val(v,s)` , where v is an array and s is a scalar. The function should return the closest value to s, that is in array v. You can assume there's only one. <br><br>
for example, given the input: v = [1,3,9] ; s=14 <br>
the function should return 1

```python
def farthest_val(v,s):
  """
  v: 1-d array (vector)
  s: scalar (number)

  Function returns the farthest value in vector v, to scalar s
  """
  farthest= v[0]
  for num in v:
    if np.abs(num-s)>np.abs(farthest-s):
      farthest=num
  return farthest
```

## 3. Thresholding
Given a matrix `M` and a scalar threshold `thresh`, the function should return a new matrix with same dimensions as `M`, that contains:

-   2 where values is equal or above threshold
-   -2 where values are below threshold

for example `thresh=3`,
$M=\begin{pmatrix}
1 & 2 & 9\\
-1 & 4 & 0\\
\end{pmatrix}$
<br>
output will be:<br>
$Out=\begin{pmatrix}
-2 & -2 & 2\\
-2 & 2 & -2\\
\end{pmatrix}$
<br><br>
Hint: be cautious not to overwrite the original M matrix

```python
def thresh_func(M,thresh) -> np.ndarray:
  """
  M: 2-d array (matrix)
  thresh: scalar (number)
  Function returns a new matrix with all values in M that are less than thresh set to -2, and values greater than thresh set to 2
  """
  mCopy= M.copy()
  for rowIndex in range(len(mCopy)):
    for cellIndex in range(len(mCopy[rowIndex])):
      if mCopy[rowIndex][cellIndex] >= thresh:
        mCopy[rowIndex][cellIndex]=2
      else:
         mCopy[rowIndex][cellIndex]=-2
  return mCopy
```

```python
#Test your code
M = np.array([[1,2,9],[-1,4,0]])
masked_arr = thresh_func(M,3)
masked_arr
```

For a given matrix M, where each `row` index is a `student` and `columns` indexes are `exercises`:<br>
calculate for the first **n** students and range of execrcises  [**a**,**b**] (b>a): <br>
* Average score for each student
* std for each student
* Maximal score of each **Exercise**
<br><br>
Example of input:
$M=\begin{pmatrix}
100 & 82 & 91\\
95 & 73 & 62\\
97 & 84 & 78\\
100 & 91 & 86\\
\end{pmatrix}$
<br>
In this matrix there are 4 students (rows) and 3 exercises (columns). <br>
The i,j element of the function is the grade of exercise j for student i<br>
for example, $M_{1,2} = 62$ : That means that student 1 got a grade of 62 for exercise 2
<br><br>


### Instructions
Complete the function `calc_grade_atts(grades,n,a,b)` Where: <br>
**grades** - input matrix (same as M mentioned before) <br>
**n** - first number of students to include in the calculation <br>
**a,b** - range of exercises to include in the calculation (b should be greater than a)<br>
<br><br>
The function should return a dictionary that contains the following:
- avg_student_grage - average grade of each student
- std_student - standart deviation of each student in relevant data
- max_grade - maximal grade for each exercise in relevant data
- **all numbers should be rounded to 2 decimal places**


relevant data - first n students and exercises in range [a,b]
<br><br>
Assume the input is valid

```python
def calc_grade_atts(grades,n,a,b):
    ezMat=grades[:n,a:b]
    return {'avg_per_student':ezMat.mean(axis=1),
            'std_per_student':ezMat.std(axis=1),
            'max_grade_per_ex':ezMat.max(axis=1)}
```

```python
#Test your solution
M = np.array([[100,82,91],[95,73,62],[97,84,78],[100,91,86]])
n_students = M.shape[0]
n_ex = M.shape[1]
calc_grade_atts(M,n_students,0,n_ex) #analyze the full input matrix
```

### 2. Meta-metascore
Create a new rating feature that takes both the [rating](https://help.imdb.com/article/imdb/track-movies-tv/ratings-faq/G67Y87TFYYP6TWAV#calculate) and the [metascore](https://www.metacritic.com/about-metascores).<br>

1. Ignore movies with not many raters: <br>Change `rating` and `metascore` of movies with below 75,000 votes to NaN.

2. Normalize each of the rating columns ("rating" and "metascore") to be between 0 to 1. Create **new columns** for that accordingly - "norm_rating" and "norm_metascore".

3. Create a new feature (new column), called "mixed_rating" to be the multiplication of the two normalized columns:
    <br>a. By using simple Arithmetic Operation.
    <br>b. By using the `apply` function.
    
4. Show top 5 movies with highest "mixed_rating" (show all table columns)

```python
#1
df.loc[df["votes"] < 75000, ["rating","metascore"]] = np.nan
#2
df["norm_rating"]=df["rating"]/10
df["norm_metascore"]=df["metascore"]/100
#3
def get_multi(row):
    return row["norm_metascore"]*row["norm_rating"]
df["mixed_rating"]=df.apply(get_multi, axis=1)
#4
df.sort_values(by="mixed_rating", ascending=False).head(5)
```
